#### Clase Plan_de_Pensiones
### Definicion de Atributos 💣
1. idPlan
- Definición: un número o combinación de letras y números que se le asigna a cada plan para diferenciarlo de los demás en el sistema.

2. saldoAcumulado

- Definición: La cantidad total de dinero que el usuario tiene ahorrada en el plan hasta el momento

3. comisionAnual

- Definición: El porcentaje o tarifa fija que la entidad financiera cobra al año por gestionar y administrar el fondo de pensiones.

4. fechaJubilacionEstimada

- Definición: La fecha en la que se prevé que el titular cumplirá la edad legal o deseada para retirarse y poder cobrar el dinero

5. tipoFondo

- Definición: La categoría del plan según dónde se invierte el dinero


---



### Definicion de funciones 👌
1. realizarAportacion(monto)

- Definición: Permite al usuario meter más dinero a su fondo de pensiones

2. simularJubilacion(edadRetiro)

- Definición: Un cálculo matemático que proyecta cuánto dinero tendrá ahorrado el usuario al llegar a la vejez si sigue aportando al ritmo actual.

3. cambiarTipoFondo(nuevoTipo)

- Definición: Permite modificar la estrategia de inversión (el tipoFondo) si el usuario decide que quiere ser más conservador o más arriesgado con su dinero.

4. aplicarRendimiento(porcentaje)

- Definición: Actualiza el saldoAcumulado sumándole las ganancias que los mercados financieros han generado en un período de tiempo.

5. solicitarRescate(motivo, monto)

- Definición: La acción de retirar el dinero. Dado que los planes de pensiones suelen estar bloqueados, esta función evalúa si el usuario cumple las condiciones legales

```python
import datetime

# =====================================================================
# CLASE: Plan_de_pensiones
# =====================================================================
class Plan_de_pensiones:
    def __init__(self, id_plan, saldo_inicial, comision, fecha_jubila, tipo_fondo):
        # Atributos
        self.id_plan = id_plan
        self.saldo_acumulado = float(saldo_inicial)
        self.comision_anual = float(comision)
        self.fecha_jubilacion_estimada = int(fecha_jubila)  # Año de jubilación
        self.tipo_fondo = tipo_fondo                       # Conservador, Moderado, Arriesgado

    # Función 1: Realizar una aportación de dinero
    def realizar_aportacion(self, monto):
        monto_float = float(monto) # Convert monto to float once
        self.saldo_acumulado += monto_float
        print(f"\n[Éxito] Se han aportado ${monto_float:.2f} al plan {self.id_plan}.")

    # Función 2: Simular jubilación usando un "Algoritmo de IA Predictivo" básico
    def simular_jubilacion(self, edad_actual):
        anios_para_retiro = self.fecha_jubilacion_estimada - datetime.datetime.now().year

        if anios_para_retiro <= 0:
            print("\n[IA] El plan ya ha alcanzado o superado su año estimado de jubilación.")
            return

        # La "IA" define un rendimiento estimado según el tipo de fondo
        tasa_estimada = 0.0
        if self.tipo_fondo.lower() == "conservador":
            tasa_estimada = 0.04  # 4% anual
        elif self.tipo_fondo.lower() == "moderado":
            tasa_estimada = 0.07  # 7% anual
        elif self.tipo_fondo.lower() == "arriesgado":
            tasa_estimada = 0.11  # 11% anual

        # Cálculo de interés compuesto simulado por la IA
        saldo_proyectado = self.saldo_acumulado * ((1 + tasa_estimada) ** anios_para_retiro)

        print("\n" + "="*50)
        print(f"🤖 ASISTENTE IA - PROYECCIÓN DE RETIRO PARA EL PLAN {self.id_plan}")
        print("="*50)
        print(f"-> Años restantes para el retiro: {anios_para_retiro} años")
        print(f"-> Estrategia analizada por IA: Fondo {self.tipo_fondo.upper()}")
        print(f"-> Tasa de rendimiento anual estimada: {tasa_estimada*100}%")
        print(f"-> SALDO ESTIMADO AL JUBILARSE: ${saldo_proyectado:,.2f}")
        print("="*50)

    # Función 3: Cambiar la estrategia de inversión (Tipo de Fondo)
    def cambiar_tipo_fondo(self, nuevo_tipo):
        self.tipo_fondo = nuevo_tipo
        print(f"\n[Configuración] Estrategia del plan {self.id_plan} cambiada a: {nuevo_tipo.upper()}.")

    # Función 4: Aplicar rendimientos automáticos del mercado
    def aplicar_rendimiento(self, porcentaje):
        ganancia = self.saldo_acumulado * (float(porcentaje) / 100)
        self.saldo_acumulado += ganancia
        print(f"\n[Mercado] ¡Rendimientos aplicados! El plan {self.id_plan} varió un {porcentaje}%. Ganancia/Pérdida: ${ganancia:.2f}")

    # Función 5: Solicitar rescate (Retiro de dinero bajo condiciones)
    def solicitar_rescate(self, motivo, monto):
        monto = float(monto)
        motivos_validos = ["jubilacion", "desempleo", "salud"]

        if motivo.lower() in motivos_validos:
            if monto <= self.saldo_acumulado:
                self.saldo_acumulado -= monto
                print(f"\n[Rescate Autorizado] Se han retirado ${monto:.2f} por concepto de: {motivo.upper()}.")
            else:
                print(f"\n[Denegado] Saldo insuficiente. Tu saldo actual es de ${self.saldo_acumulado:.2f}")
        else:
            print(f"\n[Denegado] El motivo '{motivo}' no cumple con los requisitos de la ley para retiro anticipado.")


# =====================================================================
# PROGRAMA PRINCIPAL (MENÚ INTERACTIVO)
# =====================================================================

# Instanciando los objetos usando el nuevo nombre de la clase Plan_de_pensiones
lista_planes = [
    Plan_de_pensiones("PLAN-001", 15000, 1.2, 2045, "Moderado"),
    Plan_de_pensiones("PLAN-002", 45000, 0.8, 2038, "Conservador"),
    Plan_de_pensiones("PLAN-003", 8000, 2.0, 2055, "Arriesgado")
]

def mostrar_tabla_planes():
    print("\n" + "=" * 85)
    print(f"{'ID PLAN':<12} | {'SALDO ACTUAL':<15} | {'COMISIÓN ANUAL':<16} | {'AÑO JUBILACIÓN':<16} | {'TIPO FONDO':<15}")
    print("=" * 85)
    for plan in lista_planes:
        print(f"{plan.id_plan:<12} | ${plan.saldo_acumulado:<14,.2f} | {plan.comision_anual:<14}% | {plan.fecha_jubilacion_estimada:<16} | {plan.tipo_fondo:<15}")
    print("=" * 85)

def buscar_plan(id_buscar):
    for plan in lista_planes:
        if plan.id_plan.upper() == id_buscar.upper():
            return plan
    return None

continuar = True
print("¡Bienvenido al Simulador de Planes de Pensiones con IA para Preparatoria!")

while continuar:
    print("\n--- MENÚ DE OPCIONES ---")
    print("1. Ver lista de Planes de Pensiones (Tabla)")
    print("2. Registrar un nuevo Plan")
    print("3. Realizar una Aportación (Guardar dinero)")
    print("4. Cambiar el tipo de fondo (Estrategia de inversión)")
    print("5. Aplicar rendimientos del mercado (Simulación del tiempo)")
    print("6. Solicitar rescate de dinero (Retiro)")
    print("7. Consultar al Asistente IA (Predicción de saldo futuro)")
    print("8. Salir del programa")

    opcion = input("Selecciona una opción (1-8): ")

    if opcion == "1":
        mostrar_tabla_planes()

    elif opcion == "2":
        print("\n--- Registrar Nuevo Plan ---")
        id_p = input("Ingresa el ID del plan (Ej. PLAN-004): ")
        saldo_i = input("Ingresa el Saldo Inicial ($): ")
        comision = input("Ingresa la comisión anual (%): ")
        anio_j = input("Ingresa el año estimado de jubilación (Ej. 2050): ")
        print("Tipos de fondo disponibles: Conservador, Moderado, Arriesgado")
        tipo_f = input("Ingresa el tipo de fondo: ")

        # Creación del nuevo objeto usando Plan_de_pensiones
        nuevo_plan = Plan_de_pensiones(id_p, saldo_i, comision, anio_j, tipo_f)
        lista_planes.append(nuevo_plan)
        print(f"\n[Éxito] ¡Plan {id_p} registrado correctamente!")

    elif opcion == "3":
        mostrar_tabla_planes()
        id_b = input("Ingresa el ID del plan al que deseas aportar: ")
        plan = buscar_plan(id_b)
        if plan != None:
            monto = input("Monto a aportar ($): ")
            plan.realizar_aportacion(monto)
        else:
            print("\n[Error] El Plan de Pensiones no existe.")

    elif opcion == "4":
        mostrar_tabla_planes()
        id_b = input("Ingresa el ID del plan a modificar: ")
        plan = buscar_plan(id_b)
        if plan != None:
            print("Nuevas opciones: Conservador, Moderado, Arriesgado")
            nuevo_t = input("Selecciona el nuevo tipo de fondo: ")
            plan.cambiar_tipo_fondo(nuevo_t)
        else:
            print("\n[Error] El Plan de Pensiones no existe.")

    elif opcion == "5":
        print("\n--- Simulación de Fluctuación de Mercados Financieros ---")
        porcentaje = input("Ingresa el porcentaje de rendimiento a aplicar (ej. 5.5 o -3.2): ")
        for p in lista_planes:
            p.aplicar_rendimiento(porcentaje)

    elif opcion == "6":
        mostrar_tabla_planes()
        id_b = input("Ingresa el ID del plan del cual quieres retirar dinero: ")
        plan = buscar_plan(id_b)
        if plan != None:
            print("Motivos legales válidos: Jubilacion, Desempleo, Salud")
            motivo = input("Ingresa el motivo del rescate: ")
            monto = input("Monto a retirar ($): ")
            plan.solicitar_rescate(motivo, monto)
        else:
            print("\n[Error] El Plan de Pensiones no existe.")

    elif opcion == "7":
        mostrar_tabla_planes()
        id_b = input("Ingresa el ID del plan que la IA va a analizar: ")
        plan = buscar_plan(id_b)
        if plan != None:
            edad = input("¿Qué edad tienes actualmente?: ")
            plan.simular_jubilacion(int(edad))
        else:
            print("\n[Error] El Plan de Pensiones no existe.")

    elif opcion == "8":
        print("\n¡Gracias por usar el simulador! Hasta luego.")
        continuar = False
    else:
        print("\n[Opción Incorrecta] Por favor, elige un número del 1 al 8.")
```
